# Phase Plot of a Damped Harmonic Oscillator

Explore how the phase-space trajectory changes when you vary the initial position, initial velocity, and damping coefficient. The plot shows velocity versus position, rather than either quantity versus time.

Run the code cells from top to bottom. This notebook uses `ipywidgets` controls and standard inline Matplotlib output; it does not require the interactive Matplotlib backend.

## Model

The damped-oscillator equation is

$$
\frac{d^2x}{dt^2} + 2\gamma\frac{dx}{dt} + \omega_0^2 x = 0
$$

For this demonstration, the natural angular frequency is fixed at \(\omega_0 = 1\). The damping coefficient is restricted to \(0 \leq \gamma < \omega_0\), so the oscillator is underdamped. The phase plot shows velocity \(v = dx/dt\) on the vertical axis and position \(x\) on the horizontal axis.

```python
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display

print('NumPy:', np.__version__)
print('Matplotlib:', matplotlib.__version__ if 'matplotlib' in globals() else plt.matplotlib.__version__)
print('ipywidgets:', widgets.__version__)
print('Matplotlib backend:', plt.get_backend())
```

```python
# Fixed natural angular frequency and time array
w0 = 1.0
t = np.linspace(0.0, 100.0, 5000)

# Interactive controls
spos = widgets.FloatSlider(
    value=5.0, min=-5.0, max=5.0, step=0.1,
    description='Initial position', continuous_update=True,
    style={'description_width': 'initial'},
    layout=widgets.Layout(width='500px')
)
svel = widgets.FloatSlider(
    value=0.0, min=-5.0, max=5.0, step=0.1,
    description='Initial velocity', continuous_update=True,
    style={'description_width': 'initial'},
    layout=widgets.Layout(width='500px')
)
sgam = widgets.FloatSlider(
    value=0.01, min=0.0, max=0.999, step=0.001,
    description='Damping coefficient γ', continuous_update=True,
    style={'description_width': 'initial'},
    layout=widgets.Layout(width='500px')
)

def plot_phase(pos, vel, gam):
    # Underdamped solution, written to handle zero initial position or velocity
    # without relying on an inverse-tangent phase calculation.
    wd = np.sqrt(w0**2 - gam**2)
    x = np.exp(-gam*t) * (
        pos * np.cos(wd*t) + ((vel + gam*pos) / wd) * np.sin(wd*t)
    )
    v = np.exp(-gam*t) * (
        vel * np.cos(wd*t) - ((gam*vel + w0**2*pos) / wd) * np.sin(wd*t)
    )

    fig, ax = plt.subplots(figsize=(7, 6))
    ax.plot(x, v, color='green', linewidth=1.8)
    ax.plot([pos], [vel], marker='o', color='red', markersize=5, label='Initial state')
    ax.set_xlabel('Position, x')
    ax.set_ylabel('Velocity, v')
    ax.set_title(f'Phase Plot (γ = {gam:.3f})')
    ax.axhline(0, color='black', linewidth=0.6)
    ax.axvline(0, color='black', linewidth=0.6)
    ax.grid(True, alpha=0.3)
    ax.legend()
    ax.set_aspect('auto')
    fig.tight_layout()
    plt.show()
    plt.close(fig)

# Connect the controls to the plot
plot_output = widgets.interactive_output(
    plot_phase,
    {'pos': spos, 'vel': svel, 'gam': sgam}
)

# Reset button
reset_button = widgets.Button(description='Reset')
def reset_values(button):
    spos.value = 5.0
    svel.value = 0.0
    sgam.value = 0.01
reset_button.on_click(reset_values)

display(widgets.VBox([spos, svel, sgam, reset_button]), plot_output)
```
